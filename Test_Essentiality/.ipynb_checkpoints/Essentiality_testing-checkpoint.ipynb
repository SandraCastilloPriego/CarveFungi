```python
essentialGenes = pd.read_csv("data/essentialGenes.csv", sep='\t', header=None)
essentialGenes
```

```python
formap = pd.read_csv("data/uniprotkb_taxonomy_id_559292_2025_08_19(1).tsv",sep="\t")
formap = formap.dropna(subset=["Gene Names (ordered locus)"])
```

```python
selGenes = {}
for k,  essential in essentialGenes.iterrows():    
    print(essential[1])
    sel = formap[formap["Gene Names (ordered locus)"].str.contains(essential[1])]
    print(sel.iloc[0]["RefSeq"])
    try:
        for gen in sel.iloc[0]["RefSeq"].split(";"):
            if len(gen)< 3:
                continue
            if " [" in gen:
                gen= gen.split(" [")[0]
            gen = gen.split(".")[0]
            selGenes[gen] = essential[1]
    except:
        pass
```

```python
model= cobra.io.read_sbml_model("../CarveFungi/results/Saccharomyces cerevisiae--839.6428285154577-971.6627611556756.sbml")
model.reactions.UF02997_E.lower_bound=-1000

deletion_results = single_gene_deletion(model)
deletions_essential = deletion_results[deletion_results["growth"]==0]
predicted_essential=[]
for k, r in deletions_essential.iterrows():
    predicted_essential.append(str(list(r["ids"])[0]))
    
all_genes = []
for gene in model.genes:
    all_genes.append(gene.id)
```

```python
modelActSeek= cobra.io.read_sbml_model("../CarveFungi/results/Saccharomyces cerevisiae_ActSeekN--2459.900716071449-2881.031673918169.sbml")
modelActSeek.reactions.UF02997_E.lower_bound=-1000
deletion_results_actseek = single_gene_deletion(modelActSeek)
deletions_essential_actseek = deletion_results_actseek[deletion_results_actseek["growth"]==0]
predicted_essential_actseek=[]
for k, r in deletions_essential_actseek.iterrows():
    predicted_essential_actseek.append(str(list(r["ids"])[0]))
    
all_genes_actseek = []
for gene in modelActSeek.genes:
    all_genes_actseek.append(gene.id)
```

```python
gpruler_model= cobra.io.read_sbml_model("../data/GPR/models/yeast-GEM.xml")
deletion_results = single_gene_deletion(gpruler_model)
deletions_essential = deletion_results[deletion_results["growth"]==0]
predicted_essential_gpruler=[]
for k, r in deletions_essential.iterrows():
    predicted_essential_gpruler.append(str(list(r["ids"])[0]))
    
all_genes_gpruler = []
for gene in gpruler_model.genes:
    all_genes_gpruler.append(gene.id)
```

```python
y_true = []
y_pred = []
for gene in all_genes:
    if gene.split(".")[0] in selGenes:
        print(gene, selGenes[gene.split(".")[0]])
    if gene.split(".")[0] in selGenes and selGenes[gene.split(".")[0]] in essential_genes:
        y_true.append("N")
    else:
        y_true.append("V")
    if gene in predicted_essential:
        y_pred.append("N")
    else:
        y_pred.append("V")
```

```python
y_true_actseek = []
y_pred_actseek = []
for gene in all_genes_actseek:
    print(gene)
    if gene.split(".")[0] in selGenes:
        print(gene, selGenes[gene.split(".")[0]])
    if gene.split(".")[0] in selGenes and selGenes[gene.split(".")[0]] in essential_genes:
        y_true_actseek.append("N")
    else:
        y_true_actseek.append("V")
    if gene in predicted_essential_actseek:
        y_pred_actseek.append("N")
    else:
        y_pred_actseek.append("V")
```

```python
y_true_yeast = []
y_pred_yeast = []
for gene in all_genes_gpruler:
    print(gene)    
    if gene in essential_genes:
        y_true_yeast.append("N")
    else:
        y_true_yeast.append("V")
    if gene in predicted_essential_gpruler:
        y_pred_yeast.append("N")
    else:
        y_pred_yeast.append("V")
```

```python

import numpy as np
from sklearn.metrics import confusion_matrix

labels = ["V", "N"]

def normalized_confusion(y_true, y_pred, labels):
    cm = confusion_matrix(y_true, y_pred, labels=labels)
    cm = cm / cm.sum()  # normalize by total samples
    return cm
```

```python
cm_model  = normalized_confusion(y_true, y_pred,  labels)
cm_model_actseek  = normalized_confusion(y_true_actseek, y_pred_actseek,  labels)
cm_model_yeast  = normalized_confusion(y_true_yeast, y_pred_yeast,  labels)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
def plot_cm(ax, cm, title):
    sns.heatmap(
        cm,
        ax=ax,
        annot=True,
        fmt=".3f",
        cmap="cividis",
        cbar=True,
        cbar_kws={"shrink": 0.5},
        xticklabels=["V", "N"],
        yticklabels=["V", "N"],
        square=True
    )
    ax.set_xlabel("Predictive labels")
    ax.set_ylabel("True labels")
    ax.set_title(title)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 6))

plot_cm(axes[1], cm_model, "CarveFungi yeast model")
plot_cm(axes[2], cm_model_actseek,  "CarveFungi yeast model\n with ActSeekN annotations")
plot_cm(axes[0], cm_model_yeast,  "Yeast9 model")
plt.tight_layout()

plt.savefig("GPR_essentiality_comp.svg", format="svg", dpi=300) 
plt.show()
```
